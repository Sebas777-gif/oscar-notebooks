## 1 Galois Theory

https://tube.mathe.social/w/tGevrRaSgJadGJtCUkJ8LX

Let's compute some Galois groups! We have a univariate rational polynomial.

```julia
Qx, x = QQ["x"]
f = x^4+4*x^2+7
```

How can we find its Galois group? We first look at the roots of the polynomial, on which the Galois group acts.

```julia
roots(f)
```

Hmm, the roots are not necessarily rational, so we need a field extension. Let's try the complex numbers!

```julia
roots(ComplexField(), f)
```

Even better: computing $p$-adically will result in faster convergence!

```julia
roots(padic_field(31; precision = 5), f)
```

And now we're ready to calculate our Galois group!

```julia
G, C = galois_group(f; prime = 31)
describe(G)
```

## 2 Ehrhart Theory

https://tube.mathe.social/w/nfRPyed3BGt8EX1jEysjhx

Let's count some lattice points! Here is a three-dimensional cube with its vertices having $\pm 1$ coordinates.

```julia
P = cube(3)
```

Is this indeed a lattice polytope?

```julia
is_lattice_polytope(P)
```

Yes! So let's count the number of lattice points.

```julia
lattice_points(P)
```

But what if we scale the cube and count again? That's where we can use the Ehrhart polynomial!

```julia
f = ehrhart_polynomial(P)
```

Plugging in $1$ yields our original count of $27$:

```julia
f(1)
```

Now let's plug in $2$ to see how scaling by two affects the number of points.

```julia
f(2)
```

That is the number of lattice points of the cube having vertex coordinates $\pm 2$. Discrete geometry, neatly encoded!
